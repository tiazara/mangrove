# 🌏 Ringkasan Variabel Penelitian: *Coastal Squeeze* Mangrove Pantura

Dokumen ini merangkum **10 variabel** yang digunakan untuk mengukur fenomena *coastal squeeze* — kondisi di mana mangrove "terjepit" antara tekanan laut yang naik dan rintangan buatan manusia di darat — yang dikelompokkan ke dalam **3 dimensi utama**.

---

## 🌊 Dimensi 1 — Fisik / Hazard
*Tekanan ekstrem dari arah laut yang mendorong mangrove agar tidak bisa bertahan di posisinya saat ini.*

| # | Variabel | Satuan | Sumber Data | Fungsi / Kegunaan |
|---|----------|--------|-------------|--------------------|
| 1 | **Laju Amblesan Tanah** (*Land Subsidence Rate*) | cm/tahun | Data GNSS Zenodo (Susilo et al., *Nature Scientific Data*, 2023) & Geodetik ESDM | Mengukur kecepatan permukaan tanah tenggelam di bawah muka air laut. Contoh: Sayung–Demak amblas 14–16 cm/th, Pekalongan 11,59 cm/th. **Variabel paling mematikan** karena merendam akar napas mangrove secara permanen. |
| 2 | **Laju Perubahan Garis Pantai** (*Shoreline Change Rate* / Abrasi) | m/tahun | Vektor Garis Pantai OSM / BIG & CoastSat | Mengukur kecepatan mundurnya bibir pantai akibat hantaman ombak dan hilangnya sedimen laut. |
| 3 | **Kenaikan Muka Laut Relatif** (*Relative Sea Level Rise*) | mm/tahun | Altimetri Satelit AVISO / Copernicus Marine | Mengukur kenaikan muka laut absolut yang, jika digabung dengan amblesan tanah, menghasilkan laju genangan rob. |

---

## 🌿 Dimensi 2 — Ekologis
*Kondisi kesehatan dan riwayat mangrove — seberapa rapuh dan terancam hutan saat ini.*

| # | Variabel | Satuan | Sumber Data | Fungsi / Kegunaan |
|---|----------|--------|-------------|--------------------|
| 4 | **Laju Kehilangan Luasan Mangrove Historis** | %/tahun atau ha | Global Mangrove Watch (GMW v3.0: 1996, 2010, 2020) & data statistik Excel | Membuktikan tren riil penyusutan mangrove (Demak–Semarang kehilangan 77,49 ha) dan tingkat fragmentasi sabuk mangrove. |
| 5 | **Kerapatan Kanopi Hijau** (NDVI) | Indeks (-1 s/d 1) | Sentinel-2 MSI Level-2A (Band 8 NIR & Band 4 Red) | Menilai lebat atau gundulnya tajuk kanopi mangrove di tiap segmen transek saat ini. |
| 6 | **Stres Genangan & Kelembapan Tajuk** (NDMI) | Indeks (-1 s/d 1) | Sentinel-2 MSI Level-2A (Band 8 NIR & Band 11 SWIR) | Mendeteksi stres fisiologis pohon mangrove akibat salinitas berlebih dan genangan air laut berkepanjangan. |

---

## 🧱 Dimensi 3 — Antropogenik
*Rintangan keras di arah darat — menentukan apakah mangrove punya ruang untuk mundur/migrasi ke darat saat laut naik, atau justru terbentur "tembok" buatan manusia (**inland barrier**).*

| # | Variabel | Satuan | Sumber Data | Fungsi / Kegunaan |
|---|----------|--------|-------------|--------------------|
| 7 | **Proporsi Hambatan Tambak** (*Aquaculture Barrier*) | 0–1 | ESA WorldCover 10m (Kelas 80) | Menghitung seberapa banyak area di belakang mangrove yang sudah disekat menjadi tambak ikan/udang berpematang keras (~70% lanskap pesisir Demak). |
| 8 | **Proporsi Hambatan Terbangun** (*Built-up Barrier*) | 0–1 | ESA WorldCover 10m (Kelas 50) | Menghitung area yang sudah menjadi "dinding mati" bagi mangrove: permukiman padat, pabrik, jalan raya, tanggul beton pelabuhan. |
| 9 | **Elevasi & Kemiringan Lereng** (*Accommodation Space*) | meter & derajat | Copernicus DEM 30m / DEMNAS | Mengukur topografi daratan. Daratan landai & elevasi rendah → mangrove teoretis bisa mundur ke darat. Jika curam/terhalang tanggul → ruang akomodasi alami = 0 (*terjepit total*). |
| 10 | **Kepadatan Penduduk Pesisir** | jiwa/piksel 100m | WorldPop Indonesia 2020 | Menilai besarnya tekanan aktivitas manusia dan eksploitasi air tanah di kawasan penyangga pesisir. |

## 1. `stations_information.txt` — daftar stasiun GNSS (folder `7775016`)
Folder `7775016` adalah ID record Zenodo (Susilo et al., 2023): 20 stasiun CORS di Jawa.

````python
stations = pd.read_csv(DATA / "7775016/stations_information.txt", sep=r"\s+")
stations = stations.rename(columns={"#stationName": "station"})
print(stations.shape)
stations
````

## 2. `.rneu` — time series GNSS North/East/Up (meter)
Kolom: tanggal, tahun desimal, perubahan North/East/Vertical terhadap posisi referensi, dan error masing-masing.
File `.rneu` di root `X:/Dataset` (CJPR, CPKL, CSEM) identik dengan salinannya di folder `7775016`.

````python
RNEU_COLS = ["yyyymmdd", "year_dec", "north", "east", "vert", "n_err", "e_err", "v_err"]

def read_rneu(path):
    df = pd.read_csv(path, sep=r"\s+", comment="#", header=None, names=RNEU_COLS)
    df["date"] = pd.to_datetime(df["yyyymmdd"].astype(str), format="%Y%m%d")
    return df

csem = read_rneu(DATA / "7775016/CSEM_clean_Noff1site.rneu")
display(csem.head())
csem.describe().T
````

````python
# Ringkasan semua stasiun: rentang waktu dan tren vertikal linier (indikatif, belum memperhitungkan musiman/outlier)
rows = []
for p in sorted((DATA / "7775016").glob("*.rneu")):
    d = read_rneu(p)
    slope = np.polyfit(d["year_dec"], d["vert"], 1)[0]
    rows.append({
        "station": p.name[:4], "n_obs": len(d),
        "start": d["date"].min().date(), "end": d["date"].max().date(),
        "vert_trend_cm_per_yr": round(slope * 100, 2),
    })
rneu_summary = pd.DataFrame(rows).merge(stations[["station", "Lon", "Lat"]], on="station")
rneu_summary.sort_values("vert_trend_cm_per_yr")
````

````python
fig, axes = plt.subplots(3, 1, figsize=(10, 7), sharex=True)
for ax, col in zip(axes, ["north", "east", "vert"]):
    ax.plot(csem["date"], csem[col] * 100, ".", ms=2)
    ax.set_ylabel(f"{col} (cm)")
axes[0].set_title("CSEM (Semarang) — time series NEU")
plt.tight_layout()
````

## 3. `.pos` — time series posisi GNSS format PBO (IGb14)
Header deskripsi kolom ada di atas; baris data dimulai setelah baris yang diawali `*YYYYMMDD`.
Berisi koordinat XYZ, lat/lon/height, dN/dE/dU beserta standar deviasi & korelasi.

````python
def read_pos(path):
    lines = Path(path).read_text().splitlines()
    header_idx = next(i for i, l in enumerate(lines) if l.startswith("*YYYYMMDD"))
    cols = lines[header_idx].lstrip("*").split()
    df = pd.read_csv(path, sep=r"\s+", skiprows=header_idx + 1, header=None, names=cols,
                     dtype={"YYYYMMDD": str, "HHMMSS": str})
    df["date"] = pd.to_datetime(df["YYYYMMDD"], format="%Y%m%d")
    return df

print("\n".join(Path(DATA / "7775016/CSEM.sub.final_igb14.pos").read_text().splitlines()[:9]))
csem_pos = read_pos(DATA / "7775016/CSEM.sub.final_igb14.pos")
print(csem_pos.shape)
csem_pos.head()
````

## 8. `CSPI_Raster_Variables_Pantura.tif` — raster multi-band variabel Pantura
Nodata disimpan sebagai `NaN` (bukan nilai `nodata` di metadata), jadi statistik memakai fungsi `nan*`.

````python
TIF = DATA / "CSPI_Raster_Variables_Pantura.tif"
with rasterio.open(TIF) as src:
    print("ukuran  :", src.width, "x", src.height, "| band:", src.count, "| dtype:", src.dtypes[0])
    print("crs     :", src.crs, "| resolusi:", src.res)
    print("bounds  :", src.bounds)
    print("nodata  :", src.nodata)
    band_names = list(src.descriptions)
    stack = src.read()

band_stats = []
for name, band in zip(band_names, stack):
    valid = band[~np.isnan(band)]
    band_stats.append({
        "band": name, "valid_px": valid.size, "pct_nan": round(100 * (1 - valid.size / band.size), 2),
        "min": valid.min() if valid.size else np.nan, "max": valid.max() if valid.size else np.nan,
        "mean": valid.mean() if valid.size else np.nan, "unique_(<=5)": np.unique(valid)[:5] if valid.size else [],
    })
pd.DataFrame(band_stats)
````

````python
fig, axes = plt.subplots(2, 4, figsize=(16, 6))
for ax, name, band in zip(axes.flat, band_names, stack):
    im = ax.imshow(band[::4, ::4], cmap="viridis")
    ax.set_title(name); ax.axis("off")
    fig.colorbar(im, ax=ax, shrink=0.7)
axes.flat[-1].axis("off")
plt.tight_layout()
````
